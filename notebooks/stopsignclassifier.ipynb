```python
# Files : (https://media.newyorker.com/photos/62f54954e77a7c7222469cc4/master/w_2240,c_limit/shouts-stop-sign.jpg) ( https://media.geeksforgeeks.org/wp-content/uploads/20250323153743680534/recognize_stop_sign.zip )

import cv2
from matplotlib import pyplot as plt

img = cv2.imread("/content/shouts-stop-sign.webp")
```

```python
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
plt.imshow(img_rgb)
plt.show()
```

```python
stop_cascade = cv2.CascadeClassifier('stop_data.xml')
```

```python
found = stop_cascade.detectMultiScale(img_gray, minSize=(20, 20))
```

```python
for (x, y, w, h) in found:
    cv2.rectangle(img_rgb, (x, y), (x + w, y + h), (0, 255, 0), 5)
```

```python
plt.imshow(img_rgb)
plt.show()
```
